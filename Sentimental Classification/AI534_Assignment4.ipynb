# 2.1 - Sentence Embedding and k-NN

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

train_data = pd.read_csv('train.csv')
dev_data = pd.read_csv('dev.csv')
test_data = pd.read_csv('test.csv')

X_train = train_data['sentence']
y_train = train_data['target']

X_dev = dev_data['sentence']
y_dev = dev_data['target']

X_test = test_data['sentence']
```

```python
def sentence_embedding(sentence, wv):
    words = sentence.split()
    word_vectors = [wv[word] for word in words if word in wv]
    if word_vectors:
        return np.mean(word_vectors, axis=0)
    else:
        return np.zeros(wv.vector_size)

X_train_embeddings = np.array([sentence_embedding(sent, wv) for sent in X_train])
X_dev_embeddings = np.array([sentence_embedding(sent, wv) for sent in X_dev])
X_test_embeddings = np.array([sentence_embedding(sent, wv) for sent in X_test])
```

# Question 1

```python
from scipy.spatial.distance import cdist

first_sentence_embedding = X_train_embeddings[0]

distances = cdist([first_sentence_embedding], X_train_embeddings, metric='cosine')

closest_index = np.argsort(distances[0])[1] 

closest_sentence = X_train.iloc[closest_index]
closest_label = y_train.iloc[closest_index]

print("First sentence in the training set:", X_train.iloc[0])
print("Closest sentence in the training set:", closest_sentence)
print("Label of the closest sentence:", closest_label)
```

# Question 2

```python
second_sentence_embedding = X_train_embeddings[1]

distances = cdist([second_sentence_embedding], X_train_embeddings, metric='cosine')

closest_index = np.argsort(distances[0])[1] 

closest_sentence = X_train.iloc[closest_index]
closest_label = y_train.iloc[closest_index]

print("Second sentence in the training set:", X_train.iloc[1])
print("Closest sentence in the training set:", closest_sentence)
print("Label of the closest sentence:", closest_label)
```

# Question 3

```python
k_values = range(1, 100, 2)
error_rates = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_embeddings, y_train)

    y_dev_pred = knn.predict(X_dev_embeddings)
    error_rate = 1 - accuracy_score(y_dev, y_dev_pred)
    error_rates.append(error_rate)
    print(f"k={k}, Dev Error Rate: {error_rate}")

best_k = k_values[np.argmin(error_rates)]
print(f"Best k: {best_k}, Best Error Rate: {min(error_rates)}")
```

```python
best_knn = KNeighborsClassifier(n_neighbors=best_k)
best_knn.fit(X_train_embeddings, y_train)

y_test_pred = best_knn.predict(X_test_embeddings)

output = pd.DataFrame({'id': test_data['id'], 'sentence': test_data['sentence'], 'target': y_test_pred})
output.to_csv('test.predicted.knn.csv', index=False)

print("Predictions saved to test.predicted.knn.csv")
```

```python

def tokenize(sentences):
    return [sentence.split() for sentence in sentences]

tokenized_sentences = tokenize(X_train)

vocabulary = {word for sentence in tokenized_sentences for word in sentence}
word_to_index = {word: idx for idx, word in enumerate(vocabulary)}

def one_hot_encoding(sentences, word_to_index):
    word_vectors = []
    for sentence in sentences:
        vector = np.zeros(len(word_to_index))
        for word in sentence.split():
            if word in word_to_index:
                vector[word_to_index[word]] += 1
        word_vectors.append(vector)
    return np.array(word_vectors)

X_train_one_hot = one_hot_encoding(X_train, word_to_index)
X_dev_one_hot = one_hot_encoding(X_dev, word_to_index)
X_test_one_hot = one_hot_encoding(X_test, word_to_index)
```

```python
k_values = range(1, 100, 2)
error_rates_one_hot = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_one_hot, y_train)

    y_dev_pred = knn.predict(X_dev_one_hot)
    error_rate = 1 - accuracy_score(y_dev, y_dev_pred)
    error_rates_one_hot.append(error_rate)
    print(f"k={k}, Dev Error Rate (One-Hot): {error_rate}")

best_k_one_hot = k_values[np.argmin(error_rates_one_hot)]
print(f"Best k (One-Hot): {best_k_one_hot}, Error Rate: {min(error_rates_one_hot)}")
```

```python
best_knn_one_hot = KNeighborsClassifier(n_neighbors=best_k_one_hot)
best_knn_one_hot.fit(X_train_one_hot, y_train)

y_test_pred_one_hot = best_knn_one_hot.predict(X_test_one_hot)

output_one_hot = pd.DataFrame({'id': test_data['id'],'sentence': test_data['sentence'], 'target': y_test_pred_one_hot})
output_one_hot.to_csv('test.predicted.onehot.knn.csv', index=False)

print("Predictions saved to test.predicted.onehot.knn.csv")
```

```python
from __future__ import division
import numpy as np
import pandas as pd
import time
from gensim.models import KeyedVectors


wv = KeyedVectors.load("embs_train.kv")  

def read_from(textfile):
    data = pd.read_csv(textfile)
    for i in range(len(data)):
        id, words, label = data.iloc[i]
        yield (1 if label == "+" else -1, words.split())

def make_word_vectors_embedding(sentence, wv):
    words = sentence.split()
    vectors = [wv[word] for word in words if word in wv]
    if vectors:
        return np.mean(vectors, axis=0)
    else:
        return np.zeros(wv.vector_size)

def test(devfile, model, bias, wv):
    total, err = 0, 0
    for i, (label, words) in enumerate(read_from(devfile), 1):
        sentence_embedding = make_word_vectors_embedding(" ".join(words), wv)
        prediction = np.dot(model, sentence_embedding) + bias
        err += label * prediction <= 0
    return err / i

def train(trainfile, devfile, epochs, wv):
    t = time.time()
    best_err = 1.0
    model = np.zeros(wv.vector_size)  
    bias = 0  
    best_model = model.copy()
    best_bias = bias

    for it in range(1, epochs + 1):
        updates = 0
        for i, (label, words) in enumerate(read_from(trainfile), 1):
            sentence_embedding = make_word_vectors_embedding(" ".join(words), wv)
            prediction = np.dot(model, sentence_embedding) + bias
            if label * prediction <= 0:
                updates += 1
                model += label * sentence_embedding
                bias += label
        dev_err = test(devfile, model, bias, wv)
        if dev_err < best_err:
            best_err = dev_err
            best_model = model.copy()
            best_bias = bias
        print(f"epoch {it}, updates {updates}, dev err {dev_err:.4f}")
    print(f"best dev err {best_err:.4f}, time: {time.time() - t:.2f} secs")
    return best_model, best_bias

trainfile = "train.csv"
devfile = "dev.csv"
epochs = 10

best_model, best_bias = train(trainfile, devfile, epochs, wv)
```

```python
def predict_test(testfile, model, bias, wv, output_file="test.predicted.basic.perceptron.csv"):
    test_data = pd.read_csv(testfile)
    predictions = []

    for _, row in test_data.iterrows():
        sentence = row['sentence']  
        sentence_embedding = make_word_vectors_embedding(sentence, wv)
        prediction = np.dot(model, sentence_embedding) + bias
        predictions.append("+" if prediction > 0 else "-")

    output_df = pd.DataFrame({
        'id': test_data['id'],  
        'sentence': test_data['sentence'], 
        'target': predictions
    })
    output_df.to_csv(output_file, index=False)
    print(f"Predictions saved to {output_file}")

testfile = "test.csv"
output_file = "test.predicted.basic.perceptron.csv"

predict_test(testfile, best_model, best_bias, wv, output_file=output_file)
```

```python
from __future__ import division
import numpy as np
import pandas as pd
import time
from gensim.models import KeyedVectors


wv = KeyedVectors.load("embs_train.kv") 

def precompute_embeddings(datafile, wv, is_test=False):
    data = pd.read_csv(datafile)
    embeddings = []
    labels = [] if not is_test else None

    for _, row in data.iterrows():
        sentence = row['sentence']
        embedding = make_word_vectors_embedding(sentence, wv)
        embeddings.append(embedding)
        if not is_test:
            label = 1 if row['target'] == "+" else -1
            labels.append(label)

    embeddings = np.array(embeddings)
    if is_test:
        return data, embeddings
    else:
        return data, embeddings, np.array(labels)

def make_word_vectors_embedding(sentence, wv):
    words = sentence.split()
    vectors = [wv[word] for word in words if word in wv]
    if vectors:
        return np.mean(vectors, axis=0)
    else:
        return np.zeros(wv.vector_size)

def test_dense(embeddings, labels, model):
    predictions = np.dot(embeddings, model)
    errors = np.sum(labels * predictions <= 0)
    return errors / len(labels)

def train_averaged_perceptron(train_embeddings, train_labels, dev_embeddings, dev_labels, epochs=10):
    t = time.time()
    best_err = 1.0
    model = np.zeros(train_embeddings.shape[1]) 
    aux_vector = np.zeros(train_embeddings.shape[1])  
    total_examples = 0

    for it in range(1, epochs + 1):
        updates = 0
        for idx in range(len(train_labels)):
            embedding = train_embeddings[idx]
            label = train_labels[idx]
            prediction = np.dot(model, embedding)
            if label * prediction <= 0:
                updates += 1
                model += label * embedding
                aux_vector += total_examples * label * embedding
            total_examples += 1

        averaged_model = model - (aux_vector / total_examples)

        dev_err = test_dense(dev_embeddings, dev_labels, averaged_model)
        best_err = min(best_err, dev_err)
        print(f"epoch {it}, updates {updates}, dev err {dev_err:.4f}")

    print(f"best dev err {best_err:.4f}, time: {time.time() - t:.2f} secs")
    return averaged_model

trainfile = "train.csv"
devfile = "dev.csv"
testfile = "test.csv"

train_data, train_embeddings, train_labels = precompute_embeddings(trainfile, wv)
dev_data, dev_embeddings, dev_labels = precompute_embeddings(devfile, wv)
test_data, test_embeddings = precompute_embeddings(testfile, wv, is_test=True)

epochs = 10
best_averaged_model = train_averaged_perceptron(train_embeddings, train_labels, dev_embeddings, dev_labels, epochs)
```

```python
def predict_test(test_data, test_embeddings, model, output_file="test.predicted.averaged.perceptron.csv"):
    predictions = np.dot(test_embeddings, model)
    predictions = ["+" if pred > 0 else "-" for pred in predictions]

    output_df = pd.DataFrame({
        'id': test_data['id'],  
        'sentence': test_data['sentence'],  
        'target': predictions
    })
    output_df.to_csv(output_file, index=False)
    print(f"Predictions saved to {output_file}")

output_file = "test.predicted.averaged.perceptron.csv"
predict_test(test_data, test_embeddings, best_averaged_model, output_file=output_file)
```

```python
from __future__ import division
import numpy as np
import pandas as pd
import time
from gensim.models import KeyedVectors

wv = KeyedVectors.load("embs_train.kv")  

def make_embedding_vector_with_pruning(sentence, wv, word_counts, min_freq):
    words = sentence.split()
    vectors = [
        wv[word]
        for word in words
        if word in wv and word_counts.get(word, 0) >= min_freq
    ]
    if vectors:
        return np.mean(vectors, axis=0)
    else:
        return np.zeros(wv.vector_size)

def precompute_embeddings_with_pruning(datafile, wv, word_counts, min_freq, is_test=False):
    data = pd.read_csv(datafile)
    embeddings = []
    labels = [] if not is_test else None

    for _, row in data.iterrows():
        sentence = row['sentence']
        embedding = make_embedding_vector_with_pruning(sentence, wv, word_counts, min_freq)
        embeddings.append(embedding)
        if not is_test:
            label = 1 if row['target'] == "+" else -1
            labels.append(label)

    embeddings = np.array(embeddings)
    if is_test:
        return data, embeddings
    else:
        return data, embeddings, np.array(labels)

def get_word_frequencies(filename):
    data = pd.read_csv(filename)
    word_counts = {}
    for sentence in data['sentence']:
        for word in sentence.split():
            word_counts[word] = word_counts.get(word, 0) + 1
    return word_counts

def test_dense_with_bias(embeddings, labels, model, bias):
    predictions = np.dot(embeddings, model) + bias
    errors = np.sum(labels * predictions <= 0)
    return errors / len(labels)

def train_averaged_perceptron_with_pruning(train_embeddings, train_labels, dev_embeddings, dev_labels, epochs=10):
    t = time.time()
    best_err = 1.0
    model = np.zeros(train_embeddings.shape[1]) 
    aux_vector = np.zeros(train_embeddings.shape[1])  
    bias = 0  
    aux_bias = 0 
    total_examples = 0

    for it in range(1, epochs + 1):
        updates = 0
        for idx in range(len(train_labels)):
            embedding = train_embeddings[idx]
            label = train_labels[idx]
            prediction = np.dot(model, embedding) + bias
            if label * prediction <= 0:
                updates += 1
                model +=  label * embedding
                bias +=  label
                aux_vector += total_examples * label * embedding
                aux_bias += total_examples * label
            total_examples += 1

        averaged_model = model - (aux_vector / total_examples)
        averaged_bias = bias - (aux_bias / total_examples)

        dev_err = test_dense_with_bias(dev_embeddings, dev_labels, averaged_model, averaged_bias)
        best_err = min(best_err, dev_err)
        print(f"epoch {it}, updates {updates}, dev err {dev_err:.4f}")

    print(f"best dev err {best_err:.4f}, time: {time.time() - t:.2f} secs")
    return averaged_model, averaged_bias
```

```python
trainfile = "train.csv"
devfile = "dev.csv"
testfile = "test.csv"
min_freq = 2  
epochs = 10

word_counts = get_word_frequencies(trainfile)

train_data, train_embeddings, train_labels = precompute_embeddings_with_pruning(trainfile, wv, word_counts, min_freq)
dev_data, dev_embeddings, dev_labels = precompute_embeddings_with_pruning(devfile, wv, word_counts, min_freq)
test_data, test_embeddings = precompute_embeddings_with_pruning(testfile, wv, word_counts, min_freq, is_test=True)

best_perceptron_pruned_model, best_perceptron_pruned_bias = train_averaged_perceptron_with_pruning(
    train_embeddings, train_labels, dev_embeddings, dev_labels, epochs
)
```

```python
def predict_test_with_bias(test_data, test_embeddings, model, bias, output_file):
    predictions = np.dot(test_embeddings, model) + bias
    predictions = ["+" if pred > 0 else "-" for pred in predictions]

    output_df = pd.DataFrame({
        'id': test_data['id'],
        'sentence': test_data['sentence'],
        'target': predictions
    })
    output_df.to_csv(output_file, index=False)
    print(f"Predictions saved to {output_file}")
```

```python
output_file = "test.predicted.pruned.perceptron.csv"
predict_test_with_bias(test_data, test_embeddings, best_perceptron_pruned_model, best_perceptron_pruned_bias, output_file)
```

```python
from sklearn.linear_model import LogisticRegression
import time
import numpy as np
import pandas as pd
from gensim.models import KeyedVectors

def load_embeddings(file_path):
    return KeyedVectors.load(file_path)

def compute_sentence_embeddings(data_path, embeddings):
    data = pd.read_csv(data_path)
    sentences = data["sentence"]
    labels = data["target"].apply(lambda x: 1 if x == "+" else -1)

    vectors = [
        np.mean([embeddings[word] for word in sentence.split() if word in embeddings], axis=0)
        if any(word in embeddings for word in sentence.split()) 
        else np.zeros(embeddings.vector_size)
        for sentence in sentences
    ]
    
    return np.array(vectors), labels.to_numpy()

def compute_test_embeddings(data_path, embeddings):
    data = pd.read_csv(data_path)
    sentences = data["sentence"]
    ids = data["id"]

    vectors = [
        np.mean([embeddings[word] for word in sentence.split() if word in embeddings], axis=0)
        if any(word in embeddings for word in sentence.split()) 
        else np.zeros(embeddings.vector_size)
        for sentence in sentences
    ]
    
    return np.array(vectors), ids, sentences

def main():
    embeddings = load_embeddings("embs_train.kv")
    
    train_features, train_labels = compute_sentence_embeddings("train.csv", embeddings)
    dev_features, dev_labels = compute_sentence_embeddings("dev.csv", embeddings)

    start_time = time.time()
    model = LogisticRegression(max_iter=1000)
    model.fit(train_features, train_labels)

    dev_predictions = model.predict(dev_features)
    dev_error_rate = 1 - np.mean(dev_predictions == dev_labels)
    elapsed_time = time.time() - start_time
    
    print(f"Development Error Rate: {dev_error_rate:.2f}")
    print(f"Execution Time: {elapsed_time:.2f} seconds")

    test_features, test_ids, test_sentences = compute_test_embeddings("test.csv", embeddings)
    test_labels = model.predict(test_features)

    results = pd.DataFrame({
        "id": test_ids,
        "sentence": test_sentences,
        "target": ["+" if label == 1 else "-" for label in test_labels]
    })
    output_file = "test.predictions.logistic.regression.csv"
    results.to_csv(output_file, index=False)
    print(f"Predictions saved to the file named: {output_file}")

if __name__ == "__main__":
    main()
```

# Implemented Logistic Regression + Sentence Transformer Models + GridSearchCV

```python
import pandas as pd
import numpy as np
import time

from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

def parse_labeled_data(csv_path):
    try:
        data = pd.read_csv(csv_path)
        texts = data['sentence'].tolist()
        labels = [1 if str(t).strip() == '+' else -1 for t in data['target']]
        return texts, labels
    except Exception as e:
        print(f"Error parsing {csv_path}: {e}")
        return [], []

def parse_test_data(csv_path):
    try:
        data = pd.read_csv(csv_path)
        texts = data['sentence'].tolist()
        ids = data['id'].tolist()
        return ids, texts
    except Exception as e:
        print(f"Error parsing {csv_path}: {e}")
        return [], []


model_names = [
    "sentence-transformers/all-MiniLM-L6-v2",
    "sentence-transformers/all-mpnet-base-v2",
    "sentence-transformers/paraphrase-MiniLM-L3-v2"
]

best_dev_error = float('inf')
best_model_name = None
best_model = None

train_texts, train_labels = parse_labeled_data('train.csv')
dev_texts, dev_labels = parse_labeled_data('dev.csv')

# Trying different embedding models
for model_name in model_names:
    start_time = time.time()
    
    embedder = SentenceTransformer(model_name)
    
    print(f"Computing embeddings with {model_name}...")
    train_embeddings = embedder.encode(train_texts, show_progress_bar=True)
    dev_embeddings = embedder.encode(dev_texts, show_progress_bar=True)
    
    param_grid = {
        'C': [0.001, 0.01, 0.1, 1, 10, 100],
        'penalty': ['l2'],
        'class_weight': [None, 'balanced']
    }
    
    lr = LogisticRegression(
        max_iter=2000, 
        solver='lbfgs',
        multi_class='auto'
    )
    
    grid_search = GridSearchCV(
        lr, 
        param_grid, 
        scoring='accuracy', 
        cv=5,  
        verbose=1
    )

    grid_search.fit(train_embeddings, train_labels)

    best_model_in_search = grid_search.best_estimator_
    
    dev_predictions = best_model_in_search.predict(dev_embeddings)
    dev_error_rate = 1 - accuracy_score(dev_labels, dev_predictions)
    
    if dev_error_rate < best_dev_error:
        best_dev_error = dev_error_rate
        best_model_name = model_name
        best_model = best_model_in_search

    print(f"Model: {model_name}")
    print(f"Best Parameters: {grid_search.best_params_}")
    print(f"Dev Error Rate: {dev_error_rate:.4f}")
    
    elapsed_time = time.time() - start_time
    print(f"Execution Time: {elapsed_time:.2f} seconds\n")

print(f"Best Overall Model: {best_model_name}")
print(f"Best Dev Error Rate: {best_dev_error:.4f}")

test_ids, test_sentences = parse_test_data('test.csv')
print("Computing embeddings for test...")

best_embedder = SentenceTransformer(best_model_name)
test_embeddings = best_embedder.encode(test_sentences, show_progress_bar=True)

test_preds = best_model.predict(test_embeddings)

results = pd.DataFrame({
    "id": test_ids,
    "sentence": test_sentences,
    "target": ["+" if label == 1 else "-" for label in test_preds]
})

output_file = "test.predicted.refined_sentiment_classifier_output.csv"
results.to_csv(output_file, index=False)
print(f"Output saved to {output_file}")
```
